# MMS data analysis

Download MMS FGM magnetic field and FPI ion/electron moments with pySPEDAS. Install the optional tools first with `pip install -e ".[mms]"`.

```python
from mms_data_analysis import load_mms_data, plot_mms_data, summarize_data

START = "2015-10-16 13:06:00"
END = "2015-10-16 13:07:00"
PROBE = 1
MODE = "auto"  # auto prefers burst, then falls back to fast
```

```python
data = load_mms_data(START, END, probe=PROBE, mode=MODE)
if not data.series:
    raise RuntimeError("No MMS data found; try MODE = 'fast' or another interval.")
print(f"Loaded MMS{PROBE} {data.cadence} data.")
```

```python
summary = summarize_data(data)
summary
```

```python
figure = plot_mms_data(data)
figure.show()
```
